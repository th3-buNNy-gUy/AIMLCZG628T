Given your sensor parameters, you have a mix of continuous numerical data and a binary categorical variable ('anomaly'). This allows for a wide range of statistical tests, depending on your objectives (e.g., understanding relationships, detecting anomalies, comparing conditions).

Here's a breakdown of suggested statistical tests and what you can infer from them, categorized by common analytical goals:

**I. Understanding Relationships and Dependencies (Correlation and Regression)**

* **Tests:**
    * **Pearson Correlation Coefficient:** For assessing linear relationships between any two continuous variables (e.g., Accelerometer1RMS and Volume Flow RateRMS, Current and Temperature).
    * **Spearman Rank Correlation Coefficient:** Use if the relationships are non-linear or if your data is not normally distributed.
    * **Linear Regression:** To model the relationship between a dependent continuous variable and one or more independent continuous variables (e.g., predicting Temperature based on Current and Voltage, or predicting Accelerometer1RMS based on Volume Flow RateRMS and Pressure).
    * **Multiple Regression:** An extension of linear regression for multiple independent variables.
    * **Polynomial Regression:** If a linear relationship isn't sufficient to capture the relationship between variables.

* **Inferences:**
    * **Strength and Direction of Relationships:** How strongly and in what direction (positive or negative) two variables move together. For instance, you might find a strong positive correlation between Current and Temperature, suggesting that as current increases, so does temperature.
    * **Predictive Power:** How well one or more variables can predict another. For example, can you predict the Pressure based on the Volume Flow RateRMS?
    * **Identification of Key Drivers:** Which sensor readings have the most significant impact on others. This can help in understanding system behavior and prioritizing sensor monitoring.
    * **Early Warning Signs:** If a particular sensor reading (e.g., Accelerometer RMS) starts to correlate strongly with another (e.g., Temperature) under certain conditions, it might indicate a developing issue.

**II. Comparing Groups (Anomalous vs. Non-Anomalous States)**

* **Tests:**
    * **Independent Samples t-test:** To compare the means of a continuous variable between two independent groups (e.g., compare the mean Accelerometer1RMS when `anomaly` = 0 vs. `anomaly` = 1). Use this if your continuous data is approximately normally distributed.
    * **Mann-Whitney U Test (Non-parametric alternative to t-test):** Use if the continuous data is not normally distributed or if assumptions for the t-test are not met.
    * **ANOVA (Analysis of Variance):** While `anomaly` is binary, if you were to group your data by different *types* of anomalies (if `anomaly` had more categories), ANOVA would be used to compare means across multiple groups. For your current `anomaly` variable, it's essentially a multi-group comparison with only two groups, making the t-test (or Mann-Whitney) more direct.

* **Inferences:**
    * **Significant Differences in Sensor Readings During Anomalies:** Are there statistically significant differences in the average values of Current, Pressure, Temperatures, or Accelerometer RMS readings when an anomaly is present compared to when it's not?
    * **Characterizing Anomalies:** What are the typical sensor signatures of an anomalous state? For example, is an anomaly usually accompanied by higher Accelerometer RMS values and higher Current?
    * **Identifying Anomaly Triggers/Indicators:** Which sensor parameters are most indicative of an impending or current anomaly? This is crucial for developing anomaly detection algorithms.

**III. Anomaly Detection and Classification (Beyond simple comparison)**

* **Tests/Techniques (More advanced, but highly relevant for 'anomaly' variable):**
    * **Logistic Regression:** To model the probability of an anomaly (0 or 1) based on the continuous sensor readings. This can help identify which sensor values contribute to the likelihood of an anomaly.
    * **Decision Trees/Random Forests:** For classifying data points as anomalous or not. These methods can also show which sensor parameters are most important for the classification.
    * **Support Vector Machines (SVMs):** Another powerful classification algorithm for anomaly detection.
    * **One-Class SVM:** If you only have data for "normal" operations and want to identify deviations.
    * **Isolation Forest:** Specifically designed for anomaly detection in multi-dimensional data.
    * **Clustering Algorithms (e.g., K-Means, DBSCAN):** While not a "test" in the traditional sense, clustering can help identify natural groupings in your data. Data points that don't fit into these clusters might be anomalies.
    * **Time Series Analysis (if data is sequential):**
        * **ARIMA/SARIMA:** For forecasting sensor values and detecting anomalies based on deviations from predicted values.
        * **Control Charts (e.g., X-bar and R charts):** To monitor sensor readings over time and detect out-of-control conditions, which could indicate anomalies.

* **Inferences:**
    * **Predictive Anomaly Detection:** Can you build a model that accurately predicts if a new data point is anomalous based on its sensor readings?
    * **Identification of Anomaly Patterns:** What combination of sensor readings typically indicates an anomaly?
    * **Real-time Anomaly Alerts:** Develop systems that automatically flag potential issues based on live sensor data.
    * **Root Cause Analysis:** By understanding which sensors are most influential in anomaly classification, you can get clues about the underlying cause of the anomaly.

**IV. Data Distribution and Outlier Detection**

* **Tests/Techniques:**
    * **Descriptive Statistics:** Mean, median, mode, standard deviation, variance, range for all continuous variables.
    * **Histograms/Box Plots:** Visualizing the distribution of each continuous variable.
    * **Shapiro-Wilk Test / Kolmogorov-Smirnov Test:** To test for normality of the continuous data. This is important as many parametric tests assume normality.
    * **IQR (Interquartile Range) Method:** A simple method to identify outliers in individual continuous variables.
    * **Z-score / Grubb's Test:** For identifying outliers based on standard deviations from the mean (assuming normality).

* **Inferences:**
    * **Data Characteristics:** Understanding the central tendency, spread, and shape of your sensor data.
    * **Presence of Outliers:** Identifying extreme sensor readings that might be due to sensor errors or genuine abnormal operating conditions.
    * **Assumptions for Parametric Tests:** Knowing if your data meets the assumptions for tests like the t-test or ANOVA. If not, you'd need to use non-parametric alternatives or transform the data.

**Important Considerations and Next Steps:**

1.  **Data Quality:** Before performing any tests, ensure your data is clean. Handle missing values, outliers (decide whether to remove, impute, or treat as anomalies), and potential data entry errors.
2.  **Temporal Aspect:** You mention "point is anomalous," which suggests individual data points. However, if these sensor readings are collected over time, time-series analysis becomes incredibly powerful for detecting anomalies, forecasting, and understanding dynamic system behavior.
3.  **Domain Knowledge:** Always combine statistical findings with domain expertise. What might seem like a statistical anomaly might be a normal operating condition under specific circumstances, and vice-versa.
4.  **Hypothesis Formulation:** Before running tests, clearly define the questions you want to answer (your hypotheses). This will guide your choice of tests.
5.  **Software:** You'll need statistical software (R, Python with libraries like SciPy, NumPy, Pandas, Scikit-learn, Statsmodels; SPSS; SAS; Minitab) to perform these tests effectively.

By systematically applying these statistical tests, you can gain deep insights into your sensor data, understand the relationships between different operational parameters, effectively detect anomalies, and potentially predict failures in your system.

### Part 1: Data Loading, Initial Exploration, and Preprocessing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
from sklearn.svm import OneClassSVM
from sklearn.neighbors import LocalOutlierFactor
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# --- Configuration ---
FILE_PATH = 'sensor_data.csv' # Make sure this CSV file is in the same directory or provide full path
TARGET_COLUMN = 'anomaly'
# List of all continuous sensor columns
CONTINUOUS_SENSORS = [
    'Accelerometer1RMS', 'Accelerometer2RMS', 'Current', 'Pressure',
    'Temperature', 'Thermocouple', 'Voltage', 'VolumeFlowRateRMS'
]

# Set plot style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

print("--- Starting Sensor Data Analysis ---")

# --- 1. Data Loading ---
try:
    df = pd.read_csv(FILE_PATH)
    print(f"\nSuccessfully loaded data from {FILE_PATH}")
    print(f"Initial shape: {df.shape}")
except FileNotFoundError:
    print(f"Error: The file '{FILE_PATH}' was not found. Please ensure it's in the correct directory.")
    exit() # Exit if data cannot be loaded

# --- 2. Initial Data Inspection ---
print("\n--- 2. Initial Data Inspection ---")
print("\nFirst 5 rows:")
print(df.head())

print("\nColumn information (dtypes, non-null counts):")
df.info()

print("\nDescriptive statistics for numerical columns:")
print(df.describe().T)

print(f"\nValue counts for '{TARGET_COLUMN}' column:")
print(df[TARGET_COLUMN].value_counts())
print(f"Anomaly ratio: {df[TARGET_COLUMN].sum() / len(df):.2%}")

# --- 3. Data Preprocessing ---
print("\n--- 3. Data Preprocessing ---")

# 3.1 Handle Missing Values
print("\nChecking for missing values:")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])

if missing_values.sum() > 0:
    print("\nHandling missing values...")
    # Option 1: Impute with median (robust to outliers)
    # This is a common strategy for sensor data if missingness is not extensive
    for col in CONTINUOUS_SENSORS:
        if col in df.columns and df[col].isnull().any():
            median_val = df[col].median()
            df[col].fillna(median_val, inplace=True)
            print(f"  Filled missing values in '{col}' with median: {median_val:.2f}")

    # Option 2: Drop rows with missing 'anomaly' (if anomaly itself is missing)
    if df[TARGET_COLUMN].isnull().any():
        initial_rows = len(df)
        df.dropna(subset=[TARGET_COLUMN], inplace=True)
        print(f"  Dropped {initial_rows - len(df)} rows with missing '{TARGET_COLUMN}'.")

    print("\nMissing values after imputation/dropping:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
else:
    print("No missing values found.")


# 3.2 Outlier Detection and Handling (Initial Pass - Visualization)
# We won't remove outliers yet, but visualize them to understand their impact.
print("\n--- 4. Initial Outlier Visualization (Box Plots) ---")
plt.figure(figsize=(15, 10))
for i, col in enumerate(CONTINUOUS_SENSORS):
    if col in df.columns:
        plt.subplot(3, 3, i + 1)
        sns.boxplot(y=df[col])
        plt.title(f'Box Plot of {col}')
plt.tight_layout()
plt.suptitle('Box Plots for Continuous Sensor Data (Initial View)', y=1.02, fontsize=16)
plt.show()

print("\nPreprocessing complete. Data is ready for statistical analysis.")
```

### Part 2: Statistical Tests - Relationships and Comparisons

```python
# Assuming 'df' from Part 1 is available and preprocessed

print("\n--- Part 2: Statistical Tests - Relationships and Comparisons ---")

# --- 2.1 Correlation Analysis ---
print("\n--- 2.1 Correlation Analysis (Pearson & Spearman) ---")
print("\nPearson Correlation Matrix (Linear Relationships):")
correlation_matrix_pearson = df[CONTINUOUS_SENSORS].corr(method='pearson')
print(correlation_matrix_pearson)

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix_pearson, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Pearson Correlation Matrix of Sensor Data')
plt.show()

print("\nSpearman Correlation Matrix (Monotonic Relationships - robust to non-normality/outliers):")
correlation_matrix_spearman = df[CONTINUOUS_SENSORS].corr(method='spearman')
print(correlation_matrix_spearman)

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix_spearman, annot=True, cmap='viridis', fmt=".2f", linewidths=.5)
plt.title('Spearman Correlation Matrix of Sensor Data')
plt.show()

print("\nInferences from Correlation:")
print("- High positive correlation (e.g., > 0.7): Variables increase together.")
print("- High negative correlation (e.g., < -0.7): One variable increases as the other decreases.")
print("- Low correlation (close to 0): Little to no linear/monotonic relationship.")
print("- Look for strong correlations that might indicate dependencies (e.g., Current and Temperature).")
print("- Discrepancies between Pearson and Spearman suggest non-linear relationships or influence of outliers.")


# --- 2.2 Group Comparisons (Anomaly vs. Non-Anomaly) ---
print(f"\n--- 2.2 Group Comparisons: '{TARGET_COLUMN}' (Anomaly=1) vs. Normal (Anomaly=0) ---")

normal_data = df[df[TARGET_COLUMN] == 0]
anomaly_data = df[df[TARGET_COLUMN] == 1]

if anomaly_data.empty:
    print(f"Warning: No anomalous data points ({TARGET_COLUMN}=1) found. Skipping group comparisons.")
else:
    print(f"Normal data points: {len(normal_data)}")
    print(f"Anomalous data points: {len(anomaly_data)}")

    # Check for normality before choosing t-test vs. Mann-Whitney U
    print("\nChecking for normality (Shapiro-Wilk Test - p > 0.05 suggests normal distribution):")
    alpha = 0.05 # Significance level
    normal_dist_features = {}
    for col in CONTINUOUS_SENSORS:
        if col in normal_data.columns and not normal_data[col].empty:
            stat_normal, p_normal = stats.shapiro(normal_data[col].sample(min(len(normal_data), 500), random_state=42) if len(normal_data) > 500 else normal_data[col])
            stat_anomaly, p_anomaly = stats.shapiro(anomaly_data[col].sample(min(len(anomaly_data), 500), random_state=42) if len(anomaly_data) > 500 else anomaly_data[col])
            print(f"  {col}: Normal Data p-value={p_normal:.3f} ({'Normal' if p_normal > alpha else 'Not Normal'}) | Anomaly Data p-value={p_anomaly:.3f} ({'Normal' if p_anomaly > alpha else 'Not Normal'})")
            normal_dist_features[col] = (p_normal > alpha) and (p_anomaly > alpha)


    print("\nPerforming Independent Samples T-test or Mann-Whitney U Test:")
    print("  (Null Hypothesis: Means/medians are equal between normal and anomalous groups)")
    print("  (Alternative Hypothesis: Means/medians are different)")
    print("  (If p-value < 0.05, reject Null Hypothesis - significant difference)")

    results = []
    for col in CONTINUOUS_SENSORS:
        if col in df.columns:
            if normal_data[col].empty or anomaly_data[col].empty:
                print(f"  Skipping {col}: One of the groups is empty.")
                continue

            # Check for variance equality (Levene's test) for t-test
            # Not strictly required before Levene, but good practice.
            # Levene's test p > 0.05 suggests equal variances
            stat_levene, p_levene = stats.levene(normal_data[col].dropna(), anomaly_data[col].dropna())

            # Decide between t-test and Mann-Whitney U based on normality and variance
            if normal_dist_features.get(col, False) and p_levene > alpha: # Both normal and equal variances
                stat, p = stats.ttest_ind(normal_data[col].dropna(), anomaly_data[col].dropna(), equal_var=True)
                test_type = "T-test (Equal Var)"
            elif normal_dist_features.get(col, False) and p_levene <= alpha: # Both normal, but unequal variances
                stat, p = stats.ttest_ind(normal_data[col].dropna(), anomaly_data[col].dropna(), equal_var=False)
                test_type = "T-test (Unequal Var - Welch)"
            else: # Not normal or other conditions not met for t-test
                stat, p = stats.mannwhitneyu(normal_data[col].dropna(), anomaly_data[col].dropna(), alternative='two-sided')
                test_type = "Mann-Whitney U"

            mean_normal = normal_data[col].mean()
            mean_anomaly = anomaly_data[col].mean()
            median_normal = normal_data[col].median()
            median_anomaly = anomaly_data[col].median()

            results.append({
                'Sensor': col,
                'Test Type': test_type,
                'Statistic': f"{stat:.3f}",
                'P-value': f"{p:.3f}",
                'Significantly Different': 'Yes' if p < alpha else 'No',
                'Mean (Normal)': f"{mean_normal:.3f}",
                'Mean (Anomaly)': f"{mean_anomaly:.3f}",
                'Median (Normal)': f"{median_normal:.3f}",
                'Median (Anomaly)': f"{median_anomaly:.3f}"
            })

    results_df = pd.DataFrame(results)
    print(results_df.to_string())

    print("\nInferences from Group Comparisons:")
    print("- Variables with a 'Yes' in 'Significantly Different' column show a statistically significant difference in their central tendency (mean/median) between normal and anomalous states.")
    print("- This helps identify which sensor readings are most indicative of an anomaly. For example, if 'Current' has a significantly higher mean during anomalies, it's a strong anomaly indicator.")

    # Visualize differences
    print("\nVisualizing differences between Normal and Anomaly groups (Box Plots):")
    plt.figure(figsize=(15, 12))
    for i, col in enumerate(CONTINUOUS_SENSORS):
        if col in df.columns:
            plt.subplot(3, 3, i + 1)
            sns.boxplot(x=TARGET_COLUMN, y=col, data=df)
            plt.title(f'{col} by {TARGET_COLUMN}')
    plt.tight_layout()
    plt.suptitle('Sensor Value Distributions: Normal vs. Anomaly', y=1.02, fontsize=16)
    plt.show()

# --- 2.3 Distribution Analysis (Individual Sensors) ---
print("\n--- 2.3 Distribution Analysis (Histograms) ---")
plt.figure(figsize=(15, 12))
for i, col in enumerate(CONTINUOUS_SENSORS):
    if col in df.columns:
        plt.subplot(3, 3, i + 1)
        sns.histplot(df[col], kde=True, bins=30)
        plt.title(f'Distribution of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
plt.tight_layout()
plt.suptitle('Histograms of Continuous Sensor Data', y=1.02, fontsize=16)
plt.show()

print("\nInferences from Distribution Analysis:")
print("- Histograms show the shape of the data's distribution (e.g., normal, skewed, multi-modal).")
print("- Skewness or multiple modes might indicate different operating states or underlying phenomena.")
print("- Outliers can appear as long tails or isolated bars in histograms.")

print("\nStatistical tests part complete.")
```

### Part 3: Anomaly Detection and Classification Models

```python
# Assuming 'df' from Part 1 is available and preprocessed
# Assuming normal_data and anomaly_data from Part 2 are available

print("\n--- Part 3: Anomaly Detection and Classification Models ---")

# --- 3.1 Supervised Anomaly Detection (if 'anomaly' column is reliable) ---
print("\n--- 3.1 Supervised Anomaly Detection/Classification ---")
if TARGET_COLUMN not in df.columns or df[TARGET_COLUMN].nunique() < 2:
    print(f"Skipping supervised learning: '{TARGET_COLUMN}' column is missing or does not have at least two unique values.")
else:
    X = df[CONTINUOUS_SENSORS]
    y = df[TARGET_COLUMN]

    # Handle class imbalance if anomalies are rare
    # This is a critical step for anomaly detection where anomalies (1) are often a minority class.
    anomaly_count = y.sum()
    normal_count = len(y) - anomaly_count
    if anomaly_count > 0 and normal_count > 0 and (anomaly_count / normal_count < 0.2): # If less than 20% anomalies
        print(f"\nDetected class imbalance: Normal={normal_count}, Anomaly={anomaly_count}. Ratio={anomaly_count/normal_count:.2f}")
        print("Considering using techniques for imbalanced datasets (e.g., SMOTE, class_weight).")
        # For this example, we'll use `class_weight='balanced'` in Logistic Regression and RandomForest,
        # which adjusts weights inversely proportional to class frequencies.
        # For more advanced handling, consider 'imblearn' library (pip install imblearn).

    # Scale features for models sensitive to feature scales (e.g., Logistic Regression, SVM)
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

    # Split data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X_scaled_df, y, test_size=0.3, random_state=42, stratify=y)
    print(f"\nData split: Train shape {X_train.shape}, Test shape {X_test.shape}")
    print(f"Anomaly ratio in train: {y_train.sum() / len(y_train):.2%}")
    print(f"Anomaly ratio in test: {y_test.sum() / len(y_test):.2%}")

    models = {
        "Logistic Regression": LogisticRegression(random_state=42, solver='liblinear', class_weight='balanced'),
        "Decision Tree": DecisionTreeClassifier(random_state=42, class_weight='balanced'),
        "Random Forest": RandomForestClassifier(random_state=42, n_estimators=100, class_weight='balanced')
    }

    for name, model in models.items():
        print(f"\n--- Training {name} ---")
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        y_proba = model.predict_proba(X_test)[:, 1] # Probability of being an anomaly

        print(f"\nClassification Report for {name}:")
        print(classification_report(y_test, y_pred, target_names=['Normal', 'Anomaly']))

        print(f"\nConfusion Matrix for {name}:")
        cm = confusion_matrix(y_test, y_pred)
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Predicted Normal', 'Predicted Anomaly'], yticklabels=['True Normal', 'True Anomaly'])
        plt.title(f'Confusion Matrix: {name}')
        plt.show()

        # ROC Curve and AUC
        fpr, tpr, _ = roc_curve(y_test, y_proba)
        roc_auc = auc(fpr, tpr)
        plt.figure()
        plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
        plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
        plt.xlim([0.0, 1.0])
        plt.ylim([0.0, 1.05])
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'Receiver Operating Characteristic (ROC) Curve: {name}')
        plt.legend(loc="lower right")
        plt.show()

        if hasattr(model, 'feature_importances_'):
            print(f"\nFeature Importances for {name}:")
            feature_importance_df = pd.DataFrame({
                'Feature': X.columns,
                'Importance': model.feature_importances_
            }).sort_values(by='Importance', ascending=False)
            print(feature_importance_df)
            plt.figure(figsize=(10, 6))
            sns.barplot(x='Importance', y='Feature', data=feature_importance_df)
            plt.title(f'Feature Importances: {name}')
            plt.show()
        elif hasattr(model, 'coef_'):
            print(f"\nCoefficients for {name}:")
            coef_df = pd.DataFrame({
                'Feature': X.columns,
                'Coefficient': model.coef_[0]
            }).sort_values(by='Coefficient', ascending=False)
            print(coef_df)
            plt.figure(figsize=(10, 6))
            sns.barplot(x='Coefficient', y='Feature', data=coef_df)
            plt.title(f'Coefficients: {name}')
            plt.show()


    print("\nInferences from Supervised Anomaly Detection:")
    print("- **Precision, Recall, F1-score:** Evaluate model performance, especially for the 'anomaly' class (1). Recall is often crucial for anomaly detection (minimizing false negatives).")
    print("- **ROC AUC:** A higher AUC indicates better discriminatory power of the model.")
    print("- **Confusion Matrix:** Helps understand types of errors (False Positives: normal mistaken as anomaly; False Negatives: anomaly mistaken as normal).")
    print("- **Feature Importances/Coefficients:** Identify which sensor readings are most influential in predicting anomalies. These are your strongest anomaly indicators.")

# --- 3.2 Unsupervised Anomaly Detection (if no 'anomaly' column or for novel anomalies) ---
print("\n--- 3.2 Unsupervised Anomaly Detection (assuming no labels or for novel anomalies) ---")

# We'll use the entire dataset for unsupervised methods, focusing on finding "outliers"
# If you want to train only on 'normal' data, filter df first:
# X_unsupervised = df[df[TARGET_COLUMN] == 0][CONTINUOUS_SENSORS]
# For this example, we'll apply it to the whole dataset to find deviations.
X_unsupervised = df[CONTINUOUS_SENSORS]

if X_unsupervised.empty:
    print("No data available for unsupervised anomaly detection.")
else:
    # Scale data for distance-based or distribution-based methods
    scaler_unsupervised = StandardScaler()
    X_unsupervised_scaled = scaler_unsupervised.fit_transform(X_unsupervised)
    X_unsupervised_scaled_df = pd.DataFrame(X_unsupervised_scaled, columns=X_unsupervised.columns)

    # 3.2.1 Isolation Forest
    # Good for high-dimensional data, effective at isolating anomalies
    print("\n--- Isolation Forest ---")
    # 'contamination' parameter estimates the proportion of outliers in the dataset
    # You might set this based on domain knowledge or a small sample analysis.
    # Default is 'auto' which adjusts based on the training data.
    # A value like 0.01 (1%) is a common starting point if you expect few anomalies.
    iso_forest = IsolationForest(random_state=42, contamination='auto')
    df['iso_forest_anomaly_score'] = iso_forest.decision_function(X_unsupervised_scaled_df)
    df['iso_forest_anomaly'] = iso_forest.predict(X_unsupervised_scaled_df) # -1 for outliers, 1 for inliers
    df['iso_forest_anomaly'] = df['iso_forest_anomaly'].map({1: 0, -1: 1}) # Convert to 0 (normal), 1 (anomaly)

    print(f"Isolation Forest detected {df['iso_forest_anomaly'].sum()} anomalies.")
    print(f"Top 10 most anomalous points (Isolation Forest score):")
    print(df.sort_values(by='iso_forest_anomaly_score').head(10)[['iso_forest_anomaly_score'] + CONTINUOUS_SENSORS + [TARGET_COLUMN]])

    # 3.2.2 One-Class SVM
    # Useful for novelty detection; trained on 'normal' data and detects deviations
    # For OCSVM, it's generally recommended to train only on normal data.
    if not normal_data.empty:
        print("\n--- One-Class SVM (Trained on Normal Data) ---")
        scaler_ocsvm = StandardScaler()
        X_normal_scaled = scaler_ocsvm.fit_transform(normal_data[CONTINUOUS_SENSORS])

        # 'nu' parameter is an upper bound on the fraction of training errors
        # and a lower bound of the fraction of support vectors. (similar to contamination)
        ocsvm = OneClassSVM(kernel='rbf', nu=0.05) # 5% assumed contamination in normal data
        ocsvm.fit(X_normal_scaled)

        # Predict on the entire dataset
        X_all_scaled = scaler_ocsvm.transform(df[CONTINUOUS_SENSORS])
        df['ocsvm_anomaly'] = ocsvm.predict(X_all_scaled) # -1 for outliers, 1 for inliers
        df['ocsvm_anomaly'] = df['ocsvm_anomaly'].map({1: 0, -1: 1}) # Convert to 0 (normal), 1 (anomaly)

        print(f"One-Class SVM detected {df['ocsvm_anomaly'].sum()} anomalies.")
        print(f"First 10 OCSVM anomalies and their original anomaly status:")
        print(df[df['ocsvm_anomaly'] == 1][[TARGET_COLUMN] + CONTINUOUS_SENSORS].head(10))
    else:
        print("Skipping One-Class SVM as no 'normal' data points were identified.")

    # 3.2.3 Local Outlier Factor (LOF)
    # Measures the local deviation of density of a given data point with respect to its neighbors.
    print("\n--- Local Outlier Factor (LOF) ---")
    lof = LocalOutlierFactor(n_neighbors=20, contamination='auto', novelty=False) # novelty=False for traditional LOF
    df['lof_anomaly_score'] = lof.fit_predict(X_unsupervised_scaled_df) # -1 for outliers, 1 for inliers
    df['lof_anomaly_score'] = df['lof_anomaly_score'].map({1: 0, -1: 1}) # Convert to 0 (normal), 1 (anomaly)

    print(f"LOF detected {df['lof_anomaly_score'].sum()} anomalies.")
    print(f"First 10 LOF anomalies and their original anomaly status:")
    print(df[df['lof_anomaly_score'] == 1][[TARGET_COLUMN] + CONTINUOUS_SENSORS].head(10))


    # Comparison of Unsupervised Methods with Labeled Anomalies (if available)
    if TARGET_COLUMN in df.columns and df[TARGET_COLUMN].sum() > 0:
        print("\n--- Comparing Unsupervised Anomaly Detection with Labeled Anomalies ---")
        print("\nIsolation Forest vs. Labeled Anomalies:")
        print(confusion_matrix(df[TARGET_COLUMN], df['iso_forest_anomaly']))
        print(classification_report(df[TARGET_COLUMN], df['iso_forest_anomaly'], target_names=['Normal', 'Anomaly']))

        if 'ocsvm_anomaly' in df.columns:
            print("\nOne-Class SVM vs. Labeled Anomalies:")
            print(confusion_matrix(df[TARGET_COLUMN], df['ocsvm_anomaly']))
            print(classification_report(df[TARGET_COLUMN], df['ocsvm_anomaly'], target_names=['Normal', 'Anomaly']))

        print("\nLOF vs. Labeled Anomalies:")
        print(confusion_matrix(df[TARGET_COLUMN], df['lof_anomaly_score']))
        print(classification_report(df[TARGET_COLUMN], df['lof_anomaly_score'], target_names=['Normal', 'Anomaly']))


    print("\nInferences from Unsupervised Anomaly Detection:")
    print("- These methods are useful when you don't have labeled anomaly data or when looking for novel types of anomalies.")
    print("- **Isolation Forest:** Good at finding global outliers and scales well to high dimensions.")
    print("- **One-Class SVM:** Effective for novelty detection; trained on 'normal' data to define a boundary, then detects points outside this boundary.")
    print("- **LOF:** Identifies anomalies based on local density deviations, good for finding anomalies in clusters.")
    print("- Comparison with `anomaly` column (if present) helps evaluate how well these unsupervised methods align with your predefined anomalies.")
    print("- Tuning parameters like `contamination` (or `nu` for OCSVM, `n_neighbors` for LOF) is crucial for performance.")


print("\n--- End of Analysis Script ---")
```

  * **Different Anomaly Detection Algorithms:**

      * DBSCAN (density-based clustering, anomalies are noise points).

  * **Statistical Power Analysis:** Determine the minimum sample size needed to detect an effect of a given size with a certain level of confidence.
  * **Missing Data Imputation Strategies:** Explore more sophisticated methods like K-Nearest Neighbors (KNN) imputation or MICE (Multiple Imputation by Chained Equations).
  * **Feature Engineering:** Create new features from existing ones (e.g., ratio of Accelerometer1RMS to Accelerometer2RMS, power consumption = Current \* Voltage, change in Temperature over time).
  * **Domain-Specific Thresholds:** Integrate domain knowledge to set specific thresholds for anomaly alerts based on physical limits of the system.
  * **Deployment Considerations:** If this is for a real-time system, consider efficiency, model serialization, and data streaming.
